```python
tokenizer = train_tokenizer('bpe', vocab_size=16000)
```

```python
from collections import Counter

from datasets import concatenate_datasets

imdb = load_dataset("stanfordnlp/imdb")
texts = concatenate_datasets([imdb[split] for split in ("train", "test", "unsupervised")])["text"]

token_counts = Counter()
for encoding in tokenizer.encode_batch(texts):
    token_counts.update(encoding.tokens)

print(f"{len(texts)} reviews, {sum(token_counts.values())} tokens, {len(token_counts)} distinct")
```

```python
import pandas as pd

pd.DataFrame(token_counts.most_common(100), columns=["token", "count"])
```
